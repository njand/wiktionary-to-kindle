```python
import os
import re
from collections import defaultdict
```

```python
def find_duplicate_idx_entries(directory: str):
    """
    Scan all .xhtml files in the given directory for <idx:entry id="...">
    and print duplicates.
    """
    pattern = re.compile(r'<idx:entry[^>]*\bid=["\']([^"\']+)["\']', re.IGNORECASE)
    ids = defaultdict(list)

    for root, _, files in os.walk(directory):
        for fname in files:
            if fname.lower().endswith(".xhtml"):
                path = os.path.join(root, fname)
                with open(path, "r", encoding="utf-8") as f:
                    for lineno, line in enumerate(f, start=1):
                        match = pattern.search(line)
                        if match:
                            ids[match.group(1)].append((path, lineno))

    duplicates = {k: v for k, v in ids.items() if len(v) > 1}
    return duplicates
```

```python
dupes = find_duplicate_idx_entries("outputs/KindleDictionary-la (13)/files")
if not dupes:
    print("No duplicate <idx:entry> ids found.")
else:
    for entry_id, occurrences in dupes.items():
        print(f"Duplicate id: {entry_id}")
        for path, lineno in occurrences:
            print(f"  {path}:{lineno}")
```

```python
def renumber_idx_entries(directory: str):
    """
    Normalize <idx:entry ... id="..."> across all .xhtml files:
    - Strip underscore and trailing numbers
    - Reassign sequential numbering globally per base id
    - Guarantee uniqueness across all files
    - Stable order: files sorted by path, entries by appearance
    """
    entry_pattern = re.compile(r'(<idx:entry[^>]*\bid=["\'])([^"\']+)(["\'])', re.IGNORECASE)
    base_pattern = re.compile(r'^(.*?)(?:_\d+)?$')

    # Step 1: Collect all .xhtml files in stable order
    xhtml_files = []
    for root, _, files in os.walk(directory):
        for fname in files:
            if fname.lower().endswith(".xhtml"):
                xhtml_files.append(os.path.join(root, fname))
    xhtml_files.sort()

    # Step 2: Read and collect entries in order
    file_contents = {}
    occurrences = []  # (path, start_pos, old_id, base)
    for path in xhtml_files:
        with open(path, "r", encoding="utf-8") as f:
            content = f.read()
        file_contents[path] = content

        for match in entry_pattern.finditer(content):
            old_id = match.group(2)
            base = base_pattern.match(old_id).group(1)
            occurrences.append((path, match.start(), old_id, base))

    # Step 3: Assign new ids per base, sequentially by appearance
    counters = defaultdict(int)
    new_ids = []  # parallel to occurrences
    for _, _, _, base in occurrences:
        new_id = f"{base}_{counters[base]}"
        new_ids.append(new_id)
        counters[base] += 1

    # Step 4: Rewrite files with replacements
    idx = 0
    for path in xhtml_files:
        content = file_contents[path]

        def repl(match):
            nonlocal idx
            new_id = new_ids[idx]
            idx += 1
            return f"{match.group(1)}{new_id}{match.group(3)}"

        new_content = entry_pattern.sub(repl, content)

        if new_content != content:
            with open(path, "w", encoding="utf-8") as f:
                f.write(new_content)
            print(f"Updated {path}")
```

```python
renumber_idx_entries("outputs/KindleDictionary-la (9)/files")
```

```python
from pathlib import Path

def find_entries_with_form(directory: str, form: str):
    """
    Scan all .xhtml files in the given directory for <idx:iform value="form" ...>
    and yield matching <idx:entry> blocks containing it.
    """
    entry_start = re.compile(r'<idx:entry\b', re.IGNORECASE)
    entry_end = re.compile(r'</idx:entry>', re.IGNORECASE)
    form_pattern = re.compile(
        rf'<idx:iform[^>]*\bvalue=["\']{re.escape(form)}["\']',
        re.IGNORECASE,
    )
    lemma_pattern = re.compile(
        rf'<idx:orth[^>]*\bvalue=["\']{re.escape(form)}["\']',
        re.IGNORECASE
    )

    for root, _, files in os.walk(directory):
        for fname in files:
            if fname.lower().endswith(".xhtml"):
                path = Path(root) / fname
                with open(path, "r", encoding="utf-8") as f:
                    inside_entry = False
                    buffer = []
                    for line in f:
                        if entry_start.search(line):
                            inside_entry = True
                            buffer = [line]
                        elif inside_entry:
                            buffer.append(line)
                            if entry_end.search(line):
                                entry_html = "".join(buffer)
                                if form_pattern.search(entry_html) or lemma_pattern.search(entry_html):
                                    yield path, entry_html
                                inside_entry = False
                                buffer = []
```

```python
directory = "outputs/KindleDictionary-la (15)/files"
form = "reducis"
for file, entry in find_entries_with_form(directory, form):
    print(f"Found in {file.name}:\n{entry}\n")
```
